import pandas as pd

```python
import requests
url = "https://api.openweathermap.org/data/2.5/weather"

list_cities = []
cities = [
    "Cairo",
    "Giza",
    "Alexandria",
    "Port Said",
    "Suez",
    "Ismailia",
    "Mansoura",
    "Tanta",
    "Luxor",
    "Aswan"
]
for city in cities:
    
    print(city)

    params = {
    "appid": "6a13e61e853763292101962a026aa3e8",
    "q": city
    }
    response = requests.get(url, params=params)
    data = response.json()
    city = city
    country = 'Egypt'
    lat = data['coord']['lat']
    lon = data['coord']['lon']
    temp = data['main']['temp']
    pressure = data['main']['pressure']
    Humidity = data['main']['humidity']
    wind_speed = data['wind']['speed']
    list_cities.append([city, country, lat, lon, temp, pressure, Humidity, wind_speed])



print(list_cities)
```

```python
import pandas as pd
df = pd.DataFrame(list_cities, columns=['City', 'Country', 'Latitude', 'Longitude', 'Temperature', 'Pressure', 'Humidity', 'Wind Speed'])
df
```

```python
def fun1(city):
  if city in ["Luxor", "Aswan"]:
    return 'Upper'
  else:
    return 'Lower'
df['Region'] = df['City'].apply(fun1)
df
```

```python
df['Temperature'] = df['Temperature'] - 270
df
```

```python
def categ(city):
    for city in df['City']:
        return 'Hot'
df['Temperature Category'] = df['City'].apply(categ)
df
  
```

```python
def hum(Humidity):
    if Humidity < 30:
        return 'Dry'
    elif Humidity >=30 & Humidity <=60:
        return 'Comfortable'
    else:
        return 'Humid'

df['Humidity Category'] = df['Humidity'].apply(hum)
df
```
